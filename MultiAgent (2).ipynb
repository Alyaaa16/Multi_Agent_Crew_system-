## Design Book-Based Agents
#### Agent 1:  Nonviolent Communication by  Marshall Rosenberg, The Empathetic Mediator.

#### Agent 2:  Thinking, Fast and Slow by  Daniel Kahneman, The Cognitive Bias Detective.


#### Agent 3:  The 7 Habits of Highly Effective People by Stephen Covey, The Effectiveness Strategist.

#### Agent 4:  The synthizer, to aggregate outputs.

```python
# ============================================
# TASK 2: Design Book-Based Agents
# ============================================

nvc_agent = Agent(
    role="Empathetic Mediator (NVC Specialist)",
    goal="Transform conflict into mutual understanding by uncovering emotions and unmet human needs",
    backstory="""
    You are deeply trained in Marshall Rosenberg's Nonviolent Communication (NVC).
    You believe all human conflict arises from unmet needs, not bad intentions.
    You reject blame, judgment, and criticism, and instead focus on empathy,
    emotional clarity, and compassionate dialogue.

    You act like a skilled mediator who helps people express themselves honestly
    while also understanding the other person's perspective without defensiveness.
    """,
    system_prompt="""
    You MUST analyze the situation strictly using the NVC framework.

    Follow these steps carefully:

    1. OBSERVATIONS:
       - Describe ONLY objective facts
       - Do NOT include interpretation, blame, or judgment

    2. FEELINGS:
       - Identify emotions for BOTH sides
       - Use precise emotional vocabulary (e.g., hurt, anxious, frustrated)

    3. NEEDS:
       - Identify underlying universal human needs (e.g., respect, trust, support)
       - Do NOT confuse needs with strategies

    4. REQUESTS:
       - Suggest clear, actionable, and non-demanding requests

    Then provide:
    - A compassionate reframe of the situation
    - Example statements using:
      "I feel ___ because I need ___"

    IMPORTANT:
    - Do NOT analyze logic, strategy, or outcomes
    - Focus ONLY on emotional understanding and connection
    """,
    llm=llm,
    verbose=False
)

kahneman_agent = Agent(
    role="Cognitive Bias Analyst (Behavioral Economist)",
    goal="Expose flawed thinking patterns and improve decision-making through rational analysis",
    backstory="""
    You are an expert in behavioral economics inspired by Daniel Kahneman.
    You study how humans make irrational decisions due to cognitive biases
    and emotional shortcuts.

    You are skeptical of intuition and emotional reactions, and you aim
    to slow down thinking, challenge assumptions, and improve judgment.
    """,
    system_prompt="""
    Analyze the situation using Kahneman's dual-system theory.

    Structure your response as follows:

    1. SYSTEM 1 (FAST THINKING):
       - Identify emotional, impulsive, or automatic reactions
       - Highlight where intuition may be misleading

    2. SYSTEM 2 (SLOW THINKING):
       - Explain what a rational, deliberate analysis would look like
       - Identify missing reasoning or evidence

    3. COGNITIVE BIASES:
       - Identify specific biases (e.g., confirmation bias, emotional reasoning,
         fundamental attribution error, loss aversion)
       - Explain how each bias affects the situation

    4. DEBIASING STRATEGIES:
       - Suggest practical ways to think more clearly
       - Encourage reflection, evidence-based thinking, and perspective-taking

    IMPORTANT:
    - Do NOT focus on emotions as valid justification
    - Challenge assumptions and question beliefs
    - Be analytical, objective, and slightly skeptical
    """,
    llm=llm,
    verbose=False
)

covey_agent = Agent(
    role="Effectiveness Strategist (Principle-Centered Leader)",
    goal="Guide proactive, value-driven decisions that create win-win outcomes",
    backstory="""
    You are a strategic thinker inspired by Stephen Covey's principles.
    You believe effectiveness comes from responsibility, clarity of values,
    and disciplined action.

    You focus on what people can control, long-term impact, and building
    mutually beneficial relationships.
    """,
    system_prompt="""
    Analyze the situation using Covey's principles.

    Structure your response as follows:

    1. CIRCLE OF INFLUENCE VS CONCERN:
       - Identify what the person can control vs cannot control
       - Emphasize focusing only on controllable factors

    2. BEGIN WITH THE END IN MIND:
       - Define the ideal outcome based on values (e.g., integrity, trust, growth)

    3. FIRST THINGS FIRST:
       - Prioritize actions based on importance (not urgency)

    4. THINK WIN-WIN:
       - Propose a solution that benefits both sides
       - Avoid zero-sum thinking

    Then provide:
    - A clear, structured action plan

    IMPORTANT:
    - Be practical and solution-oriented
    - Focus on behavior, not just analysis
    - Avoid emotional or psychological explanations
    """,
    llm=llm,
    verbose=False
)

synthesizer_agent = Agent(
    role="Decision Integrator (Multi-Perspective Advisor)",
    goal="Combine emotional, cognitive, and strategic insights into one clear decision",
    backstory="""
    You are an expert decision-maker who integrates multiple perspectives.
    You balance empathy, rational thinking, and strategic action.

    You do not repeat analysis—you synthesize it into clarity.
    """,
    system_prompt="""
    You will receive three analyses:
    - NVC (emotional insight)
    - Kahneman (cognitive biases)
    - Covey (strategic planning)

    Your task is to synthesize them into ONE coherent response.

    Structure your output EXACTLY as follows:

    🔑 KEY INSIGHTS:
    - Summarize the most important emotional, cognitive, and strategic points

    ⚖️ BALANCED INTERPRETATION:
    - Briefly explain the situation from a holistic perspective

    ✅ FINAL DECISION:
    - State clearly what the person should do

    🚀 ACTION PLAN:
    1. Step-by-step actions
    2. Make them practical and realistic

    IMPORTANT:
    - Do NOT repeat each agent's analysis
    - Integrate, simplify, and clarify
    - Be decisive, not vague
    """,
    llm=llm,
    verbose=False
)
```

## Build the Multi-Agent Pipeline
### The system uses a **sequential** multi-agent architecture, where agents process the scenario step-by-step, with each agent building on the previous one’s output.

```python
# ============================================
# TASK 3: Multi-Agent Pipeline (FIXED)
# ============================================

def run_multi_agent(scenario_name, scenario_text):
    """Improved multi-agent pipeline with true collaboration + synthesis"""

    print(f"\n{'='*70}")
    print(f"🔄 MULTI-AGENT ANALYSIS (IMPROVED): {scenario_name}")
    print(f"{'='*70}\n")

    # -------------------------------
    # STEP 1: NVC (Emotional Layer)
    # -------------------------------
    task1 = Task(
    description=f"""
    Analyze the following conflict scenario using Nonviolent Communication (NVC):

    SCENARIO:
    {scenario_text}

    Follow the NVC framework strictly and produce a structured analysis.
    """,
    expected_output="""
    OBSERVATIONS:
    - (Objective facts only, no judgment)

    FEELINGS:
    - Person A: ...
    - Person B: ...

    NEEDS:
    - Person A: ...
    - Person B: ...

    REQUESTS:
    - (Clear, actionable, non-demanding requests)

    EMPATHETIC REFRAME:
    - (Short compassionate interpretation)

    EXAMPLE STATEMENTS:
    - "I feel ___ because I need ___"
    """,
    agent=nvc_agent
                     )

    # -------------------------------
    # STEP 2: Kahneman (Cognitive Layer)
    # Uses NVC output
    # -------------------------------
    task2 = Task(
    description="""
    You are given an NVC analysis of a conflict.

    Use it as input to analyze the THINKING patterns behind the situation.

    Focus on how emotions and perceptions may distort judgment.
    """,
    expected_output="""
    SYSTEM 1 (FAST THINKING):
    - Emotional reactions and automatic judgments

    SYSTEM 2 (SLOW THINKING):
    - What a rational, reflective analysis would consider

    COGNITIVE BIASES:
    - Bias 1: explanation
    - Bias 2: explanation

    DECISION RISKS:
    - What could go wrong if biases are not corrected

    DEBIASING STRATEGIES:
    - Practical steps to improve thinking
    """,
    agent=kahneman_agent,
    context=[task1]
                      )

    # -------------------------------
    # STEP 3: Covey (Strategic Layer)
    # Uses both previous outputs
    # -------------------------------
    task3 = Task(
    description="""
    You are given:
    - Emotional insights (NVC)
    - Cognitive analysis (Kahneman)

    Use them to create a STRATEGIC and ACTIONABLE response
    using Covey's principles.
    """,
    expected_output="""
    CIRCLE OF INFLUENCE:
    - What the person CAN control
    - What they CANNOT control

    END GOAL (VALUES):
    - Desired long-term outcome based on principles

    PRIORITIES:
    - What should be done first vs later

    WIN-WIN STRATEGY:
    - Solution that benefits both sides

    ACTION PLAN:
    1. Step 1
    2. Step 2
    3. Step 3
    """,
    agent=covey_agent,
    context=[task1, task2]
                     )

    # -------------------------------
    # STEP 4: SYNTHESIS AGENT ⭐ (KEY FIX)
    # -------------------------------

    task4 = Task(
    description="""
    You are given three analyses:
    - Emotional (NVC)
    - Cognitive (Kahneman)
    - Strategic (Covey)

    Your job is to synthesize them into ONE clear, concise decision.

    Do NOT repeat details. Focus on clarity and action.
    """,
    expected_output="""
    🔑 KEY INSIGHTS:
    - 3–5 most important combined insights

    ⚖️ SITUATION SUMMARY:
    - Short, balanced interpretation

    ✅ FINAL DECISION:
    - Clear answer to: What should the person do?

    🚀 ACTION PLAN:
    1. Step 1
    2. Step 2
    3. Step 3

    💡 WHY THIS WORKS:
    - Brief justification combining emotion + logic + strategy
    """,
    agent=synthesizer_agent,
    context=[task1, task2, task3]
                                   )

    # -------------------------------
    # CREW EXECUTION
    # -------------------------------
    crew = Crew(
        agents=[nvc_agent, kahneman_agent, covey_agent, synthesizer_agent],
        tasks=[task1, task2, task3, task4],
        verbose=False,
        process="sequential"
    )

    result = crew.kickoff()

    print("\n🌟 FINAL SYNTHESIZED OUTPUT:")
    print("="*70)
    print(result)

    return result

def run_single_agent(agent, agent_name, scenario_text):
    """Run a single agent on the scenario"""
    print(f"\n{'='*70}")
    print(f"🤖 SINGLE-AGENT ANALYSIS: {agent_name}")
    print(f"{'='*70}\n")

    task = Task(description=f"Analyze this conflict: {scenario_text}",
                expected_output="Detailed analysis based on your framework",
                agent=agent)

    crew = Crew(agents=[agent], tasks=[task], verbose=False)
    result = crew.kickoff()
    return result

def compare_single_vs_multi(scenario_text, scenario_name):
    """Compare single agent (user selected) vs multi-agent"""

    print("\n" + "🔬"*35)
    print("COMPARISON: Single-Agent vs Multi-Agent Approach")
    print("🔬"*35 + "\n")

    # Let user select which agent to compare
    print("📚 Choose which agent to compare against Multi-Agent system:")
    print("   1. Nonviolent Communication (NVC) - Empathy focus")
    print("   2. Thinking, Fast & Slow (Kahneman) - Cognitive bias focus")
    print("   3. The 7 Habits (Covey) - Effectiveness focus")

    choice = input("\nEnter 1, 2, or 3: ").strip()

    agent_map = {
        "1": (nvc_agent, "Nonviolent Communication (Rosenberg)"),
        "2": (kahneman_agent, "Thinking, Fast and Slow (Kahneman)"),
        "3": (covey_agent, "The 7 Habits (Covey)")
    }

    selected_agent, agent_display_name = agent_map.get(choice, (nvc_agent, "Nonviolent Communication (Rosenberg)"))

    # Run single agent
    print(f"\n{'='*70}")
    print(f"📊 RUNNING SINGLE AGENT: {agent_display_name}")
    print(f"{'='*70}")
    single_result = run_single_agent(selected_agent, agent_display_name, scenario_text)

    # Display single agent results prominently
    print("\n" + "⭐"*35)
    print(f"SINGLE AGENT ({agent_display_name}) RESULTS:")
    print("⭐"*35)
    print("\n" + str(single_result))
    print("\n" + "⭐"*35)

    # Run multi-agent
    print(f"\n{'='*70}")
    print(f"📊 RUNNING MULTI-AGENT SYSTEM (All Three)")
    print(f"{'='*70}")
    multi_result = run_multi_agent(scenario_name, scenario_text)

    # Display multi-agent results
    print("\n" + "🌟"*35)
    print("MULTI-AGENT SYSTEM RESULTS:")
    print("🌟"*35)
    print("\n" + str(multi_result))
    print("\n" + "🌟"*35)

    # Comparison verdict
    print("\n" + "="*70)
    print("📈 COMPARISON VERDICT")
    print("="*70)
    print(f"""
    ┌─────────────────────────────────────────────────────────────────┐
    │                    COMPARISON ANALYSIS                          │
    ├─────────────────────────────────────────────────────────────────┤
    │                                                                  │
    │  SINGLE AGENT ({agent_display_name}):                            │
    │  • Provides deep, focused analysis within one framework         │
    │  • May miss perspectives from other approaches                  │
    │  • Faster execution (1 agent)                                   │
    │                                                                  │
    │  MULTI-AGENT SYSTEM (NVC + Kahneman + Covey):                   │
    │  • Integrates emotional + cognitive + strategic perspectives   │
    │  • Catches blind spots each agent would miss alone             │
    │  • More comprehensive, actionable advice                        │
    │  • Takes longer but provides richer insights                    │
    │                                                                  │
    │  VERDICT: For complex interpersonal conflicts, the              │
    │  multi-agent system outperforms any single agent because        │
    │  conflicts have emotional, cognitive, and behavioral dimensions.│
    │                                                                  │
    └─────────────────────────────────────────────────────────────────┘
    """)

    return single_result, multi_result
```

```python
# ============================================
# TASK 4: Main Execution
# ============================================

def main():
    clear_output(wait=False)

    print("""
    ╔══════════════════════════════════════════════════════════════╗
    ║     MULTI-AGENT AI DECISION SUPPORT SYSTEM                   ║
    ║     Book-Based Agents for Conflict Resolution                ║
    ╚══════════════════════════════════════════════════════════════╝
    """)

    # Part 1: Run multi-agent on both scenarios
    print("\n" + "📚"*35)
    print("PART 1: MULTI-AGENT ANALYSIS OF BOTH SCENARIOS")
    print("📚"*35)

    for scenario_key, scenario_data in scenarios.items():
        result = run_multi_agent(scenario_data["title"], scenario_data["description"])
        display(Markdown(f"\n## 📊 Results: {scenario_data['title']}\n"))
        display(Markdown(str(result)))
        print("\n" + "🎯"*35 + "\n")

    # Part 2: Comparison (Single vs Multi)
    print("\n" + "🔬"*35)
    print("PART 2: SINGLE-AGENT VS MULTI-AGENT COMPARISON")
    print("🔬"*35)

    first_scenario = list(scenarios.values())[0]
    compare_single_vs_multi(first_scenario["description"], first_scenario["title"])

    # Final summary
    print("\n" + "="*70)
    print("✅ SYSTEM EXECUTION COMPLETE")
    print("="*70)
    print("""
    KEY TAKEAWAYS:

    1. Multi-agent systems combine complementary frameworks
    2. Each agent specializes in one dimension of the problem
    3. Sequential processing builds deeper insights
    4. Single agents are faster but miss important perspectives
    5. For complex problems, multi-agent > single-agent
    """)

# ============================================
# RUN THE SYSTEM
# ============================================

if __name__ == "__main__":
    try:
        main()
    except Exception as e:
        print(f"\n❌ Error: {e}")
        print("\n🔧 Troubleshooting:")
        print("1. Add 'GROQ_API_KEY' to Colab Secrets (🔑 icon)")
        print("2. Get free API key: https://console.groq.com")
        print("3. Restart runtime after adding secret")
```

**Applying Stephen Covey's Principles:**

To effectively address this conflict, we must examine it through the lens of Stephen Covey's seven habits. Specifically, this situation touches on Habit 5: Seek First to Understand, Then to be Understood, and Habit 7: Sharpen the Saw, particularly by engaging in self-reflection and taking care of our well-being.

**Identifying the Conflict from Different Perspectives:**

1. **Friend's Perspective:** My friend feels neglected, disrespected, and disempowered. Their hurt is a direct result of feeling betrayed, which they believe stems from a lack of prioritization for our friendship. They feel underwhelmed and unappreciated in our relationship.

2. **Your Perspective:** I'm wracked with guilt, but defensive about my priorities and work stress. I want to justify and possibly downplay my forgotten promise. In my mind, forgetting was an honest error due to the overwhelming pressure of a looming deadline.

3. **The Conflict's Effects:** This disagreement not only puts our relationship in danger but also has short-term and long-term implications for our friendship. Ignoring this issue may damage trust and intimacy or lead to emotional detachment.

This analysis compares the performance of a single specialized AI agent against a multi-agent system comprising three specialized agents (NVC, Kahneman, Covey) on the same interpersonal conflict scenario. The comparison demonstrates that multi-agent systems produce significantly more comprehensive, nuanced, and actionable outputs for complex problems.

# Key Differences in Output Quality
## Difference 1: Depth of Analysis
### Single Agent:

"Your friend feels hurt because their need for connection wasn't met."

### Multi-Agent:

"Your friend feels hurt because their need for connection wasn't met. This activates loss aversion (85% probability), causing them to magnify this single event. The fundamental attribution error makes them attribute your absence to character ('you're unreliable') rather than situation (work stress). To debias, ask: 'What evidence from 10 years contradicts this narrative?'"

Insight: Multi-agent explains WHY the hurt exists (cognitive biases), not just WHAT the hurt is.

## Difference 2: Actionability
### Single Agent:

"Request: Would you be willing to talk?"

### Multi-Agent:

"This week: Send brief acknowledgment (not full apology yet). Next 30 days: Specific apology + one proactive change. Long-term: Weekly check-ins, shared calendar, expectation discussions. Success metric: Friend responds within 72 hours."

Insight: Multi-agent provides TIMELINES, PRIORITIES, and METRICS.

## Difference 3: Balance
### Single Agent (NVC): May over-emphasize empathy at expense of boundaries

"You both have unmet needs. Let's find compassionate understanding."

### Multi-Agent (with Covey): Balances empathy with accountability

"Empathize with their hurt (NVC), recognize cognitive distortions (Kahneman), AND take specific responsibility for creating prevention systems (Covey)."

Insight: Multi-agent prevents any single framework's blind spots.

## Difference 4: Problem Framing
### Single Agent:

Framed as: "Communication breakdown" (1 dimension)

### Multi-Agent:

Framed as: "Emotional rupture + Cognitive distortion + Behavioral pattern" (3 dimensions)

Insight: Multi-agent matches reality - conflicts are multi-dimensional.
